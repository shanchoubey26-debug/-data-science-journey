```python
import pandas as pd
```

```python
df = pd.read_csv("train.csv", encoding="latin1")
```

```python
df.head(3)
```

```python
sales_by_category = df.groupby("Category")["Sales"].sum()
sales_by_category
```

```python
sales_summary = (
    df.groupby("Category")["Sales"]
    .sum()
    .reset_index()  # Converts Category back to a regular column
    .sort_values(by="Sales", ascending=False)
)

# Format Sales to 2 decimal places
sales_summary["Sales"] = sales_summary["Sales"].round(2)
sales_summary
```

```python
# Check total sales across Category and Sub-Category
detailed_summary = (
    df.groupby(["Category", "Sub-Category"])[["Sales"]]
    .sum()
    .round(2)
    .reset_index()
)

detailed_summary.head(10)
```

```python
category_metrics = (
    df.groupby("Category")
    .agg(
        Total_Sales=("Sales", "sum"),
        Order_Count=("Order ID", "nunique"),
        Avg_Order_Value=("Sales", "mean"),
    )
    .round(2)
    .reset_index()
)

category_metrics
```
